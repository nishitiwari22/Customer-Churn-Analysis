Customer Churn Feature Analysis

Objective:
Analyze which features influence customer churn
and derive actionable business insights.

## 1. Load Data


```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
```

## 2. Dataset Overview


```python
df = pd.read_csv("../data/processed/cleaned_churn.csv")

df.head()
```

## 3. Correlation Analysis


```python
print("Shape:", df.shape)

df.info()
```

## 4. Churn by Age


```python
df["Exited"].value_counts()
```

## 5. Churn by Tenure


```python
plt.figure(figsize=(6,4))

sns.countplot(
    x="Exited",
    data=df
)

plt.title("Customer Churn Distribution")

plt.show()
```

## 6. Churn by Balance


```python
plt.figure(figsize=(12,8))

sns.heatmap(
    df.corr(),
    cmap="coolwarm",
    annot=False
)

plt.title("Feature Correlation Heatmap")

plt.show()
```

## 7. Churn by Active Membership


```python
plt.figure(figsize=(8,5))

sns.boxplot(
    x="Exited",
    y="Age",
    data=df
)

plt.title("Age vs Churn")

plt.show()
```

Observation:

Customers who churn tend to be older
than customers who remain active.

## 8. Random Forest Feature Importance


```python
plt.figure(figsize=(8,5))

sns.boxplot(
    x="Exited",
    y="Tenure",
    data=df
)

plt.title("Tenure vs Churn")

plt.show()
```

Observation:

Customers with lower tenure show
higher churn probability.

## 9. Business Insights


```python
plt.figure(figsize=(8,5))

sns.boxplot(
    x="Exited",
    y="Balance",
    data=df
)

plt.title("Balance vs Churn")

plt.show()
```

```python
sns.countplot(
    x="IsActiveMember",
    hue="Exited",
    data=df
)

plt.title("Active Membership vs Churn")

plt.show()
```
